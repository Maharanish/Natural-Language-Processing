# I. Hands-on Tools

```python
import spacy
from nltk.stem import PorterStemmer

# Load model spaCY (english)
nlp = spacy.load("en_core_web_sm")

# Input Teks
text = "Chris and Grace finally arrived at our destination and began our hike. The trail was steep and winding, but the views were worth it."
```

# 1. Sentence Splitter

```python
"""
Sentence Splitter digunakan untuk memecah teks misalnya berbentuk paragraf menjadi kalimat-kalimat terpisah.
Hal ini berguna agar setiap kalimat bisa diproses secara independen oleh pipeline NLP.
"""
text_Problem = nlp(text)
for i, sent in enumerate(text_Problem.sents, start=1):
    print(f"sentence-{i}: {sent.text}")
```

# 2. Tokenization

```python
"""
Tokenization dapat digunakan untuk memecah suatu kalimat menjadi kata-kata atau token. Dalam satu token, bisa terdiri dari hanya satu kata, gabungan dua kata, atau gabungan dari beberapa kata.
"""

# Tokenization satu kata
for token in text_Problem:
    print(token.text)
```

# 3. Stemming

```python
"""
Stemming digunakan untuk mengubah suatu kata ke bentuk dasar yang common.
Stemming biasanya menggunakan teknik normalisasi kata dengan memotong sufiks/prefiks untuk mengubah kata ke bentuk akarnya (stem) menggunakan algoritma heuristik seperti Porter.
Stemming juga biasanya menghasilkan bentuk yang bukan kata baku (contoh: “playing” → “play” atau “connection” → “connect”) dan tidak mempertimbangkan konteks morfologis.
"""
stemmer = PorterStemmer()

for token in text_Problem:
    print(token.text, "->", stemmer.stem(token.text))
```

# 4. Lemmatization

```python
"""
Lemmatization digunakan untuk mengubah suatu kata ke bentuk dasar yang valid menurut kamus (mis. “was” → “be”, “running” → “run”) dengan mempertimbangkan informasi POS dan konteks morfologis.
"""
for token in text_Problem:
    print(token.text, "->", token.lemma_)
```

# 5. Entity Masking

```python
"""
Entity Masking digunakan untuk menggantikan entitas bernama (seperti nama orang, organisasi, lokasi, tanggal, nilai uang) dalam teks dengan placeholder generik (mis. <PERSON>, <ORG>) atau pseudonim untuk tujuan anonimisasi, privasi, atau generalisasi data.
"""
entity_masking = text
for ent in text_Problem.ents:
    entity_masking = entity_masking.replace(ent.text, f"<{ent.label_}>")
print(entity_masking)
```

# 6. POS Tagger

```python
"""
POS (Part-of-speech) tagging digunakan untuk memberi label kelas kata (seperti NOUN, VERB, ADJ, ADV) pada setiap token. Tujuannya supaya komputer bisa mengerti peran setiap kata dalam kalimat.
"""

for token in text_Problem:
    print(token.text, "->", token.pos_)
```

# 7. Phrase Chunking

```python
"""
Phrase chunking (atau shallow parsing) digunakan untuk mengidentifikasi frasa dasar dalam kalimat (noun phrase).
"""
for chunk in text_Problem.noun_chunks:
    print(f"Noun Phrase :{chunk.text}")
```

# 8. Syntactic Parser

```python
"""
Syntactic Parser digunakan untuk menganalisis struktur gramatikal suatu kalimat dengan menentukan hubungan antar kata (dependency parsing).
"""
from spacy import displacy
for token in text_Problem:
    print(f"{token.text:10} | {token.dep_:10} | Head: {token.head.text:10} | POS: {token.pos_}")

# Visualisasi parse tree dengan displacy
displacy.render(text_Problem, style="dep", jupyter=True)


# Keterangan / Arti dari masing-masing simbol
# nsubj (nominal subject) → subjek kalimat, biasanya kata benda yang melakukan aksi
# Contoh: John → loves football.

# cc (coordinating conjunction) → kata hubung
# Contoh: and, or, but.

# conj (conjunct) → kata/frasa yang dihubungkan oleh cc
# Contoh: John and Mary → Mary adalah conj.

# advmod (adverbial modifier) → kata keterangan yang memodifikasi kata kerja, kata sifat, atau kata keterangan lain
# Contoh: quickly runs.

# ROOT → kata utama dari kalimat, biasanya kata kerja utama
# Contoh: runs dalam John runs fast.

# prep (prepositional modifier) → kata depan yang memodifikasi kata/frasa lain
# Contoh: in the park.

# poss (possessive modifier) → kata yang menunjukkan kepemilikan
# Contoh: John's book.

# pobj (object of preposition) → objek dari preposisi
# Contoh: in the park → park adalah pobj.

# dobj (direct object) → objek langsung dari kata kerja
# Contoh: plays football → football adalah dobj.

# punct (punctuation) → tanda baca
# Contoh: , . ! ?.

# det (determiner) → penentu kata benda
# Contoh: the park → the.

# acomp (adjectival complement) → kata sifat yang menjadi pelengkap predikat
# Contoh: He is happy → happy adalah acomp.

# npadvmod (noun phrase adverbial modifier) → frasa nominal yang berfungsi sebagai keterangan
# Contoh: last year, two days ago.
```

# II. Paper Abstract

```python
nlp = spacy.load("en_core_web_sm")
paper_abstract = "We present SemEval-2024 Task 10, a shared task centred on identifying emotions and finding the rationale behind their flips within monolingual English and Hindi-English code-mixed dialogues. This task comprises three distinct subtasks – emotion recognition in conversation for code-mixed dialogues, emotion flip reasoning for code-mixed dialogues, and emotion flip reasoning for English dialogues. Participating systems were tasked to automatically execute one or more of these subtasks. The datasets for these tasks comprise manually annotated conversations focusing on emotions and triggers for emotion shifts.1 A total of 84 participants engaged in this task, with the most adept systems attaining F1-scores of 0.70, 0.79, and 0.76 for the respective subtasks. This paper summarises the results and findings from 24 teams alongside their system descriptions."
doc = nlp(paper_abstract)
```

# Kombinasi 1

```python
TOP_N = 10
KEYWORD_POS_TAGS = ["NOUN", "VERB", "ADJ", "PROPN"]
DOMAIN_SEED_WORDS = ["dialogue", "english", "code-mixed"]
SUBTASK_KEYWORDS = ["subtask"]
DATASET_KEYWORDS = ["dataset", "annotated"]

# Lemmatization, Tokenization, Pos tagging
words = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha and not token.is_stop and token.pos_ in ["NOUN", "VERB", "ADJ", "PROPN"]
]

freq_words = {}
for word in words:
    freq_words[word] = freq_words.get(word, 0) + 1

top_words = sorted(freq_words.items(), key=lambda item: item[1], reverse=True)[:TOP_N]

# Lemmatization, Tokenization
tokens = [token.lemma_.lower() for token in doc if token.is_alpha and not token.is_stop]
bigrams = [(tokens[i], tokens[i+1]) for i in range(len(tokens)-1)]
freq_bigrams = {}
for bigram in bigrams:
    freq_bigrams[bigram] = freq_bigrams.get(bigram, 0) + 1

top_co_occurrences = sorted(freq_bigrams.items(), key=lambda item: item[1], reverse=True)[:TOP_N]

# Noun Phrase Chunking
noun_phrases = [chunk.text.lower() for chunk in doc.noun_chunks]

freq_nps = {}
for np in noun_phrases:
    freq_nps[np] = freq_nps.get(np, 0) + 1

top_noun_phrases = sorted(freq_nps.items(), key=lambda item: item[1], reverse=True)[:TOP_N]

# Result
top_words_list = [word for word, count in top_words]
keyword_candidates = [w for w in top_words_list if w in ["emotion", "flip", "reasoning", "task"]]
keyword_inti = " & ".join(set(keyword_candidates))

top_phrases_list = [phrase for phrase, count in top_noun_phrases]
domain_candidates = [np for np in top_phrases_list if any(seed in np for seed in DOMAIN_SEED_WORDS)]
domain = ", ".join(set(domain_candidates[:2]))

subtask_sents = [sent.text.strip() for sent in doc.sents if any(kw in sent.text.lower() for kw in SUBTASK_KEYWORDS)]
subtask_text = "\n".join(subtask_sents) if subtask_sents else "Not found"

dataset_sents = [sent.text.strip() for sent in doc.sents if any(kw in sent.text.lower() for kw in DATASET_KEYWORDS)]
dataset_text = dataset_sents[0] if dataset_sents else "Not found"

print("--- Top 10 Words ---")
for word, count in top_words:
    print(f"{word}: {count}")

print("\n--- Top 10 Co-occurrences (Bigrams) ---")
for (word1, word2), count in top_co_occurrences:
    print(f"('{word1}', '{word2}'): {count}")

print("\n--- Top 10 Noun Phrases ---")
for phrase, count in top_noun_phrases:
    print(f"{phrase}: {count}")

print("\n--- Final Insights ---")
print("Keyword:", keyword_inti)
print("Domain:", domain)
print("Subtask Info:", subtask_text)
print("Dataset Info:", dataset_text)
```

# Kombinasi 2

```python
# Named Entity Recognition (NER)
entities = {}
for ent in doc.ents:
    label = ent.label_
    text = ent.text
    if label not in entities:
        entities[label] = []
    entities[label].append(text)

# Phrase Chunking
key_concepts = []
for chunk in doc.noun_chunks:
    if len(chunk.text.split()) > 1 and chunk.root.pos_ != "PRON":
        key_concepts.append(chunk.text)

# Tokenization, Lemmatization
keywords = [
    token.lemma_.lower()
    for token in doc
    if not token.is_stop and not token.is_punct and token.pos_ not in ['SPACE']
]
keyword_freq_dict = {}
for keyword in keywords:
    if keyword in keyword_freq_dict:
        keyword_freq_dict[keyword] += 1
    else:
        keyword_freq_dict[keyword] = 1
sorted_keywords = sorted(keyword_freq_dict.items(), key=lambda item: item[1], reverse=True)
keyword_freq = sorted_keywords[:5]

# Syntactic Parsing
actions = []
for token in doc:
    if token.pos_ == 'VERB':
        subject, obj = '', ''
        for child in token.children:
            if child.dep_ == 'nsubj':
                subject = child.text
            if child.dep_ == 'dobj':
                obj = child.text
        if subject and obj:
            actions.append(f"'{subject}' -> '{token.lemma_}' -> '{obj}'")

# Result
print("Result")

print("\n--- 1. Important Entities ---")
for label, items in entities.items():
        print(f"  - {label}: {', '.join(set(items))}")

print("\n--- 2. Key Concepts (Main Topics) ---")
for concept in sorted(list(set(key_concepts)), key=len, reverse=True)[:5]:
        print(f"  - {concept}")

print("\n--- 3. Top 5 Most Common Keywords ---")
for keyword, freq in keyword_freq:
        print(f"  - {keyword} (appeared {freq} times)")

print("\n--- 4. Main Actions Detected ---")
for action in set(actions):
        print(f"  - {action}")
```
